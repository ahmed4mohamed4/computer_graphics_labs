## Calculation methods

| Symbol | Meaning |
|---|---|
| H | Screen height |
| W | Screen width |
| T | Video time in seconds |

### For one frame

| What we find | Formula | Example (H = 500) |
|---|---|---|
| Scan lines | H | 500 |
| Horizontal retrace | H - 1 | 499 |
| Vertical retrace | 0 | 0 |

### For a video

frame rate fixed -> `24 frames/sec`

First, find the number of frames:

**Number of frames = T × 24**



Example: T = 90 seconds, so frames = 90 × 24 = 2160

| What we find | Formula | Example (T = 90, H = 500) |
|---|---|---|
| Scan lines | frames × H | 2160 × 500 = 1,080,000 |
| Horizontal retrace | frames × (H - 1) | 2160 × 499 = 1,077,840 |
| Vertical retrace | frames - 1 | 2160 - 1 = 2159 |

## Assignment:

As explained in Lecture 2, write a Python program to calculate the number of scan lines, horizontal retraces, and vertical retraces in a raster display system. The program should take the screen width, screen height, and video duration as inputs, assuming a frame rate of 24 frames per second. Follow the calculation method illustrated in Lecture 2.

## Implementation

```python
def read_number (prompt : str, min_num : float, include_min : bool = True) -> float:
    while True:
        try:
            value = float (input (prompt))

            if include_min:
                if value < min_num:
                    raise ValueError
            else:
                if value <= min_num:
                    raise ValueError

            return value
        except ValueError:
            print (f"Enter a number greater than or equal to {min_num}")
```

```python
def frame (h : float) -> tuple:
    scan_lines = h
    horizontal_retrace = h - 1
    VERTICAL_RETRACES = 0

    return scan_lines, horizontal_retrace, VERTICAL_RETRACES
```

```python
FRAME_RATE = 24  # frames per second

def video (h : float, t : float) -> tuple:
    frames = FRAME_RATE * t
    scan_lines = frames * h
    horizontal_retrace = frames * (h - 1)
    vertical_retraces = frames - 1

    return frames, scan_lines, horizontal_retrace, vertical_retraces
```

```python
width = read_number (prompt= "Screen width (pixels): ", min_num= 1)
height = read_number (prompt= "Screen height (pixels): ", min_num= 1)
num_of_sec = read_number (prompt= "Video duration (seconds): ", min_num= 0, include_min= False)

# One frame
scan_lines, horizontal_retrace, vertical_retraces = frame (h= height)
print ("\n-> One frame")
print ("Scan lines:", scan_lines)
print ("Horizontal retrace:", horizontal_retrace)
print ("Vertical rate", vertical_retraces)


# Video
frames, scan_lines, horizontal_retrace, vertical_retraces = video (h= height, t= num_of_sec)
print ("\n-> Video")
print ("Frames:", frames)
print ("Scan lines:", scan_lines)
print ("Horizontal retrace:", horizontal_retrace)
print ("Vertical rate", vertical_retraces)
```
